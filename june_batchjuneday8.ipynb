 Set Functions

```python

#set()-to remove duplicates converts an iterable set automatically removing the duplicates
#in DS finding the unique elements
l=[1,2,3,3,4,4]
m=set(l)
print(m)

#issubset()-if a set is inside an other set
#returns a boolean value
#in DS we check for validating columns
a={1,2}
b={1,2,3}
print(a.issubset(b))#here we check if a is inside b or if a is a subset of b

#isdisjoint()-check whether if there are any overlapping elemnts or common elements
#returns a true value if there is no common elements
#in ds to check whether there is data leakage
#also to check whether there is no overlapp in multi class problems
a={1,2,3}
b={3,4,5}
print(a.isdisjoint(b))#here a and b has overlapping elements 3

#frozenset()-here the set cant be modified after creation
#suppose to we defined a set gender={"male","female","other"}
#suppose the user accidentally entered something else so there will be error
f=frozenset{1,2,3}
print(f)
#f.add(4)#here there will be the error since it cant be updated

#len(set())-count the unique values
#count the
l=[1,2,2,3,3,3,3]
num=len(set(l))
print(num)

#min()-to find the minimum value
#max()-to find the maximum value

#any()-if any elemnts satisfy the condition
#in ds check for validation ie that is if the values of one set is present in another set
s={2,3,4,5,6}
print(any(x>5 for x in s))#here checks if any elemnt is greater than 5 and we use loop

#all()-chceks if all elemnts satisfy the condition
#In data science to check for data validation
r={2,4,6,8}
print(all(x%2==0 for x in r))#checks if all elents satisfy this condition



```

write a program to find if the number is prime or not

```python
n=int(input("enter a number"))
#to check if a number to be prime it should have two factors only more than two its not prime
c=0
for i in range(1,n+1):
  if n%i==0:
    c+=1

if c==2:
    print("prime")
else:
     print("not prime")


```
