**Logical operators**


```python
#logic opeartors are used to check two or more statements
a=10
b=5
#1.AND Operators
print(a>2 and b>2)#the value returned is true or false
#here the true is returned if both statements are true
#otherwise it will be false
print(a>6 and b>6)

#2.OR Operators
print(a>2 or b>2)#here the value is returned is true or false
#here either one of the conditions should be true for the value to be true
#false is returned if both conditions are false
print(a>11 or b>11)#both statements are wrong

#3.NOT Operators
print(not a>5) #the output will be reversed
#here the output is supposed to be true but since there is not operator the value returned is false





```

**Bitwise Operators **





```python
#Bitwise AND Operator (&)
#here two numbers are taken and then AND operation is done
a=5 #Coressponding binary is 0101
b=3 #Corresponding binary is 0011
print(a&b)#The corresponding bitwise AND operation binary value is converted into decimal

#Bitwise OR Operator (|)
#here two numbers are taken and then OR operation is done
a=5 #Coressponding binary is 0101
b=3 #Corresponding binary is 0011
print(a|b)#The corresponding bitwise OR operation binary value is converted into decimal
```
