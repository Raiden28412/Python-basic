2.Trimming and padding

```python
#strip()-Removes the whitespaces

s="   lionel messi  "

a=s.strip() #stripx the whitespaces from both the sides
print(a)

b=s.lstrip() #left side removed
print(b)

c=s.rstrip() #right side removed
print(c)

#you can also remove the special character
f="####Ronaldo####"
z=f.strip("#")
print(z)


#s.center(n),s.ljust(n),s.rjust(n)-These pad the strings with spaces with the widt of n
h="CR7"
i=h.center(10) #centered
print(i)

j=h.ljust(10)#pushed to left
print(j)

k=h.rjust(10)#pushed to right
print(k)

#you can also use custom fill character instead of spaces
y=h.center(10,"*")
print(y)

#Real world scenario
print("Player Name".ljust(10),"Position".rjust(18))
print("Lionel messi".ljust(10),"Right Wing".rjust(18))


#zfill(n)-Pads the string with zeros on the left
g="42".zfill(5) #adds 3 zeros excluding the two characters already present
print(g)
```

 3.Replacing and splitting[link text](https://)

```python
#replace()-Find the occurance of of one character and replace it with other character
u="Cristiano"
x=u.replace("i","k") #replaces i with k
print(x)

#You can also limit the number of replacements
m=u.replace("i","o",1)
print(m)

#split()-Cuts the string into pieces using
v="apple,banana,mango"
l=v.split(",")
print(l)

#rsplit()-starts cutting from right side but groups left
t="a,b,c,d"
d=t.rsplit(",",2)
print(d)

#Real world use
path="home/user/documents/file.text"
new_path=path.rsplit("/",1) #cuts from right after documents
print(new_path)

#splitness()-Splits the string when there is a new line()
f="a\nb\nc\nd"
line=f.splitlines()
print(line)

#join()-takes a list of strings and join them together
l=["apple","watermelon"]
"".join(l)
print(l)
```

String slicing

```python
#extract a string from a set of characters
#two types


#forward slicing-using forawrd index
s="luminar" #extract just "lumi"
#syntax s[start:stop+1:step]
s[0:4] #here "lumi" will be printed
s[4:7] #here "nar" will be printed
s[1:] #here from u it will print the entire string
#here from luminar we need to print l m n r
#so step size will be incremented by 2
s[0::2]#step size will be incremented by 2


#backward slicing-using backward index
s="luminar"
#here the indexing will be from backward
s[-7:-3]#here print "lumi"
s[-3:-1]#here print "nar"
s[-1:4:-1]#here inorder to print "ran" here we need put step size as -1
s[-1::-1]#here print "ranimul"
s[-1:-4:-1]#here "ran" will be printed
```
