Write a program to check whether a year is leap year is not
Interview Question

```python
year=int(input("Enter a year"))
if ((year%4==0 and year%100!=0)or year%400==0):
  #for a year to be a year to be leap year it should not be divisble by 100
  #but it can be divisible by 400
  #eg:1900 is divisble by 4 but its not divisble by 100
  print("Year is leap year")

else:
  print("year is not leap year")
```

create a list of vowels from a given string

```python
s="all apples are oranges"
l=[]#empty list to append this
v="aeiouAEIOU"
for i in s:
  if i in v:
    l.append(i)#here the vowel values will be appended to the list

print(l)
```

write a program to find the number of words in a given string without using list

```python
s="all apples are oranges"
s=s.strip()#to remove the whitespaces at the beginning of the string
s=s.count(" ")
print(s+1)#no of words=whitespaces+1
```

```python
s="all apples are oranges"
c=0
for i in s:
  if i==" ":
    c+=1#assignment operation
print(c+1)#no of words=whitespaces+1
```

write a program to remove duplicate elements in a given list


```python
l=[1,1,2,2,2,2,3,3,4,5,5,9]
x=[]
for i in l:
  if i not in x:#here to check whether if the character is already in x
    x.append(i)

print(x)
```
