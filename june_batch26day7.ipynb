```python
#enumerate()-When you loop over a list normally you only get the value,
#but enumerate() gives you both the index and value
y=["apple","banana","grapes"]
for i ,item in enumerate(y):
    print(i,item)  #(i-index,value from the list coresponding to he list)
                   #item=value no mannual counter needed


#suppose you dont want your index to start displaying from 0
print("\n")
y=["apple","banana","grapes"]
for i, item in enumerate(y,start=1):
    print(i,item)

#map()-applies a function t every elemment in a list and returns a new list
#map(function,iterable)
h=[1,2,3,4]
result=map(lambda x: x**2,h)#here this map is to function is to find the square of each number in the list
square=list(result)
print(square)

#map with strings
l=["Rohan","Walter","white"]
p=map(str.upper,l)#here the string will converted to upper case
print(list(p))

#map with two lists
a1=[1,2,3]
a2=[4,5,6]
sum=map(lambda x,y:x+y,a1,a2)
print(list(sum))


#filter()-keep the elements from the filterand keeps the string where the condition is true
#filter(function,iterable)

o=[1,2,3,4,5]
t=filter(lambda x:x%2==0,o)
print(list(t))

#any()-returns true if any element is true
#all()-returns true if all elements are true
```
